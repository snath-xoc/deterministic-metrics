## IMERG

```python
from google.cloud import storage
import os

# Parameters
bucket_name = "bucket_shruti_imerg"
prefix = "6h/2025/"          # adjust if needed
download_dir = "imerg_6h_2000"

# Make sure download directory exists
os.makedirs(download_dir, exist_ok=True)

# Create an anonymous client (works if bucket is public)
client = storage.Client.create_anonymous_client()
bucket = client.bucket(bucket_name)

# List all blobs with the given prefix
blobs = bucket.list_blobs(prefix=prefix)

for blob in blobs:
    # Skip folders
    if blob.name.endswith("/"):
        continue

    # Only download files that start with '202506'
    filename = os.path.basename(blob.name)
    if filename.startswith("202506") and filename.endswith(".nc"):
        local_path = os.path.join(download_dir, filename)
        print(f"Downloading {filename}")
        blob.download_to_filename(local_path)

print("Download complete.")
```

## C-GAN

```python
from google.cloud import storage
import os

# Parameters
bucket_name = "sewaa-ifs-2025"
prefix = "24h/06/"          # the subfolder to download
download_dir = "cgan_6h_2025"

# Make sure download directory exists
os.makedirs(download_dir, exist_ok=True)

# Create an anonymous client (works if bucket is public)
client = storage.Client.create_anonymous_client()
bucket = client.bucket(bucket_name)

# List all blobs with the given prefix
blobs = bucket.list_blobs(prefix=prefix)

for blob in blobs:
    # Skip folders
    if blob.name.endswith("/"):
        continue

    # Extract relative path inside the prefix
    relative_path = os.path.relpath(blob.name, prefix)
    
    # Optional: rename to add "GAN_" prefix
    local_filename = f"GAN_{os.path.basename(relative_path)}"
    
    # Build the full local path
    local_path = os.path.join(download_dir, local_filename)
    
    # Ensure parent directories exist
    os.makedirs(os.path.dirname(local_path), exist_ok=True)
    
    print(f"Downloading {blob.name} as {local_filename}")
    blob.download_to_filename(local_path)

print("Download complete.")
```
